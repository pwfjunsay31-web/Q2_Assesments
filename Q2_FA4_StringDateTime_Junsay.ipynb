```python
first = input("Enter First Name: ").strip().title()
last = input("Enter Last Name: ").strip().title()
section = input("Enter Section: ").strip().upper()
subject = input("Enter Favorite Subject: ").strip().title()

name = first + " " + last

print("\nFormatted Output:")
print("Name:", name)
print("Name Code:", name.upper())
print("Section:", section)
print("Favorite Subject:", subject)
print("Character Count:",len(name))
```
